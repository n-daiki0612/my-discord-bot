```python
from google.colab import files
from pathlib import Path

print("推論する動画を選択してください")
uploaded_video = files.upload()

video_extensions = {
    ".mp4",
    ".avi",
    ".mov",
    ".mkv",
    ".webm",
}

video_candidates = [
    Path("/content") / filename
    for filename in uploaded_video
    if Path(filename).suffix.lower() in video_extensions
]

if len(video_candidates) != 1:
    raise RuntimeError(
        f"動画ファイルを1つだけ選択してください: {video_candidates}"
    )

VIDEO_PATH = video_candidates[0]

print("動画:", VIDEO_PATH)
print("存在:", VIDEO_PATH.is_file())
```

```python
import shutil
from pathlib import Path

LOCAL_MODEL_PATH = Path("/content/best.pt")
LOCAL_VIDEO_PATH = Path(
    f"/content/{VIDEO_PATH.name}"
)

#shutil.copy2(MODEL_PATH, LOCAL_MODEL_PATH)
#shutil.copy2(VIDEO_PATH, LOCAL_VIDEO_PATH)
LOCAL_MODEL_PATH = MODEL_PATH
LOCAL_VIDEO_PATH = VIDEO_PATH
print("モデルコピー完了:", LOCAL_MODEL_PATH)
print("動画コピー完了:", LOCAL_VIDEO_PATH)
```

```python
from pathlib import Path

import torch
from ultralytics import YOLO


DEVICE = 0 if torch.cuda.is_available() else "cpu"

OUTPUT_ROOT = Path("/content/runs/detect")
OUTPUT_NAME = "umihotaru_video_prediction1"
OUTPUT_DIR = OUTPUT_ROOT / OUTPUT_NAME

# 学習済みモデルを読み込む
model = YOLO(str(LOCAL_MODEL_PATH))

# stream=Trueではジェネレーターが返るため、
# ループして初めて動画全体が処理される
results = model.predict(
    source=str(LOCAL_VIDEO_PATH),

    # 学習時と同じ640から開始
    imgsz=640,

    # 小さい対象なのでやや低め
    conf=0.15,

    # 重複する枠を整理する基準
    iou=0.50,

    # 1フレーム内の最大検出数
    max_det=500,

    device=DEVICE,

    # 枠付き動画を保存
    save=True,

    # 全結果をRAMへ保持しない
    stream=True,

    # 全フレームを処理
    vid_stride=1,

    # 描画設定
    show_labels=True,
    show_conf=True,
    line_width=2,

    project=str(OUTPUT_ROOT),
    name=OUTPUT_NAME,
    exist_ok=True,

    verbose=False,
)

processed_frames = 0
total_detections = 0

for result in results:
    processed_frames += 1

    if result.boxes is not None:
        total_detections += len(result.boxes)

    if processed_frames % 500 == 0:
        print(
            f"{processed_frames}フレーム処理済み / "
            f"累計検出数: {total_detections}"
        )

print()
print("推論完了")
print("処理フレーム数:", processed_frames)
print("累計検出数:", total_detections)
print("保存フォルダ:", OUTPUT_DIR)
```

```python
from pathlib import Path

video_extensions = {
    ".mp4",
    ".avi",
    ".mov",
    ".mkv",
}

output_videos = sorted(
    path
    for path in OUTPUT_DIR.rglob("*")
    if path.is_file()
    and path.suffix.lower() in video_extensions
)

if not output_videos:
    raise FileNotFoundError(
        f"出力動画が見つかりません: {OUTPUT_DIR}"
    )

for path in output_videos:
    print(
        path,
        f"{path.stat().st_size / 1024**2:.1f} MB",
    )

PREDICTED_VIDEO_PATH = output_videos[0]

print()
print("使用する出力動画:", PREDICTED_VIDEO_PATH)
```

```python
import subprocess
from pathlib import Path

FINAL_VIDEO_PATH = Path(
    "/content/umihotaru_detected_000.mp4"
)

command = [
    "ffmpeg",
    "-y",
    "-i",
    str(PREDICTED_VIDEO_PATH),

    "-c:v",
    "libx264",

    "-preset",
    "fast",

    "-crf",
    "23",

    "-pix_fmt",
    "yuv420p",

    "-movflags",
    "+faststart",

    str(FINAL_VIDEO_PATH),
]

completed_process = subprocess.run(
    command,
    capture_output=True,
    text=True,
)

if completed_process.returncode != 0:
    print(completed_process.stderr)
    raise RuntimeError("MP4変換に失敗しました")

print("変換完了:", FINAL_VIDEO_PATH)
print(
    "サイズ:",
    f"{FINAL_VIDEO_PATH.stat().st_size / 1024**2:.1f} MB",
)
```
